```python
import pandas as pd
import os

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
DATA_PATH = "../data/raw/"

files = {
    "customers": "olist_customers_dataset.csv",
    "geolocation": "olist_geolocation_dataset.csv",
    "order_items": "olist_order_items_dataset.csv",
    "payments": "olist_order_payments_dataset.csv",
    "reviews": "olist_order_reviews_dataset.csv",
    "orders": "olist_orders_dataset.csv",
    "products": "olist_products_dataset.csv",
    "sellers": "olist_sellers_dataset.csv",
    "category_translation": "product_category_name_translation.csv"
}

dataframes = {}

for name, file in files.items():
    path = os.path.join(DATA_PATH, file)
    dataframes[name] = pd.read_csv(path)
    print(f"{name} chargé : {dataframes[name].shape[0]} lignes, {dataframes[name].shape[1]} colonnes")
```

## Contrôle de la qualité des données

Dans cette section, nous analysons la structure de chaque table afin d'identifier les problèmes potentiels :

- types de données incorrects ;
- valeurs manquantes ;
- doublons ;
- colonnes importantes pour l'analyse BI.

```python
for name, df in dataframes.items():
    print("\n" + "="*60)
    print(f"TABLE : {name}")
    print("="*60)
    print(df.info())
```

```python
missing_summary = []

for name, df in dataframes.items():
    missing_count = df.isna().sum()
    missing_percent = (df.isna().mean() * 100).round(2)
    
    table_missing = pd.DataFrame({
        "table": name,
        "column": missing_count.index,
        "missing_count": missing_count.values,
        "missing_percent": missing_percent.values
    })
    
    table_missing = table_missing[table_missing["missing_count"] > 0]
    missing_summary.append(table_missing)

missing_summary = pd.concat(missing_summary, ignore_index=True)
missing_summary.sort_values(by=["table", "missing_percent"], ascending=[True, False])
```

```python
duplicates_summary = []

for name, df in dataframes.items():
    duplicates_summary.append({
        "table": name,
        "rows": len(df),
        "duplicate_rows": df.duplicated().sum(),
        "duplicate_percent": round(df.duplicated().mean() * 100, 2)
    })

duplicates_summary = pd.DataFrame(duplicates_summary)
duplicates_summary
```

```python
key_checks = {
    "customers": "customer_id",
    "orders": "order_id",
    "products": "product_id",
    "sellers": "seller_id"
}

for table, key in key_checks.items():
    df = dataframes[table]
    print(f"{table} — {key}")
    print(f"Nombre de lignes : {len(df)}")
    print(f"Valeurs uniques : {df[key].nunique()}")
    print(f"Doublons sur la clé : {len(df) - df[key].nunique()}")
    print("-"*40)
```

## 3. Conversion des colonnes de dates

Les colonnes de dates sont actuellement stockées au format texte.

Pour pouvoir analyser l'évolution des ventes dans le temps, calculer les délais de livraison et créer des agrégations mensuelles, nous devons les convertir au format datetime.

```python
orders = dataframes["orders"].copy()
order_items = dataframes["order_items"].copy()
customers = dataframes["customers"].copy()
products = dataframes["products"].copy()
payments = dataframes["payments"].copy()
category_translation = dataframes["category_translation"].copy()
```

```python
orders_date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in orders_date_columns:
    orders[col] = pd.to_datetime(orders[col], errors="coerce")

orders[orders_date_columns].dtypes
```

```python
order_items["shipping_limit_date"] = pd.to_datetime(
    order_items["shipping_limit_date"],
    errors="coerce"
)

order_items["shipping_limit_date"].dtype
```

```python
orders[orders_date_columns].head()
```

### Résultat

Les colonnes de dates ont été converties avec succès au format datetime.

Cette conversion permettra de réaliser des analyses temporelles fiables, notamment l'évolution mensuelle du chiffre d'affaires et l'étude des délais de livraison.

## Création de la table principale des ventes

L'objectif de cette étape est de construire une table centrale appelée `sales`.

Cette table combine les informations nécessaires pour calculer les KPIs business :

- commandes ;
- produits vendus ;
- clients ;
- catégories ;
- paiements.

Dans cette table, une ligne représente un produit vendu dans une commande.

Cette structure est adaptée à l'analyse BI car elle permet d'étudier le chiffre d'affaires par mois, produit, catégorie, client et région.

Nous allons utiliser des jointures entre les tables principales du dataset Olist.

```python
sales = order_items.merge(
    orders,
    on="order_id",
    how="left"
)

sales = sales.merge(
    customers,
    on="customer_id",
    how="left"
)

sales = sales.merge(
    products,
    on="product_id",
    how="left"
)

sales = sales.merge(
    category_translation,
    on="product_category_name",
    how="left"
)

payments_by_order = payments.groupby("order_id", as_index=False).agg({
    "payment_value": "sum",
    "payment_installments": "max"
})

sales = sales.merge(
    payments_by_order,
    on="order_id",
    how="left"
)

sales.shape
```

### Vérification de la table centrale

Après les jointures, nous vérifions la structure de la table `sales`.

L'objectif est de contrôler :

- le nombre de colonnes ;
- les types de données ;
- la présence éventuelle de valeurs manquantes après les jointures ;
- les colonnes disponibles pour les futurs KPIs.

### Création d'une version propre de la table des ventes

La table `sales` contient toutes les colonnes obtenues après les jointures entre les commandes, les clients, les produits, les catégories et les paiements.

Pour faciliter l'analyse BI, nous créons une table simplifiée appelée `sales_clean`.

L'objectif est de garder uniquement les colonnes utiles pour répondre aux questions business :

- chiffre d'affaires total ;
- nombre de commandes ;
- panier moyen ;
- ventes par mois ;
- ventes par région ;
- ventes par catégorie ;
- clients fidèles ;
- analyse des délais de livraison.

Cette table propre servira ensuite pour l'analyse Python, les requêtes SQL et le dashboard Power BI.

```python
sales_clean = sales[
    [
        "order_id",
        "order_item_id",
        "customer_id",
        "customer_unique_id",
        "product_id",
        "seller_id",
        "order_status",
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "shipping_limit_date",
        "customer_city",
        "customer_state",
        "product_category_name",
        "product_category_name_english",
        "price",
        "freight_value",
        "payment_value",
        "payment_installments"
    ]
].copy()

sales_clean.shape
```

### Traitement des valeurs manquantes utiles

Toutes les valeurs manquantes ne doivent pas être supprimées.

Dans cette première version du projet BI, nous traitons uniquement les valeurs manquantes qui peuvent gêner directement l'analyse business.

Les catégories produits manquantes sont remplacées par `unknown`, afin de conserver les ventes associées à ces produits sans perdre de chiffre d'affaires.

```python
sales_clean["product_category_name"] = sales_clean["product_category_name"].fillna("unknown")
sales_clean["product_category_name_english"] = sales_clean["product_category_name_english"].fillna("unknown")

sales_clean[["product_category_name", "product_category_name_english"]].isna().sum()
```

### Création des colonnes utiles pour l'analyse BI

Pour faciliter les analyses SQL et Power BI, nous ajoutons plusieurs colonnes calculées.

Ces colonnes permettront de calculer rapidement les KPIs business :

- chiffre d'affaires ;
- total payé avec livraison ;
- évolution mensuelle ;
- analyse annuelle ;
- délai de livraison ;
- statut de livraison.

```python
sales_clean["revenue"] = sales_clean["price"]
sales_clean["total_value"] = sales_clean["price"] + sales_clean["freight_value"]

sales_clean["order_year"] = sales_clean["order_purchase_timestamp"].dt.year
sales_clean["order_month"] = sales_clean["order_purchase_timestamp"].dt.month
sales_clean["order_month_name"] = sales_clean["order_purchase_timestamp"].dt.month_name()
sales_clean["order_year_month"] = sales_clean["order_purchase_timestamp"].dt.to_period("M").astype(str)

sales_clean["delivery_days"] = (
    sales_clean["order_delivered_customer_date"] 
    - sales_clean["order_purchase_timestamp"]
).dt.days

sales_clean["is_delivered"] = sales_clean["order_status"] == "delivered"

sales_clean[
    [
        "price",
        "freight_value",
        "revenue",
        "total_value",
        "order_year",
        "order_month",
        "order_month_name",
        "order_year_month",
        "delivery_days",
        "is_delivered"
    ]
].head()
```

### Analyse des statuts de commande

Avant de calculer les KPIs business, nous devons vérifier les différents statuts de commande.

Cette étape permet de décider quelles commandes doivent être incluses dans l'analyse des ventes.

```python
sales_clean["order_status"].value_counts()
```

### Filtrage des commandes livrées

Pour les KPIs business principaux, nous conservons uniquement les commandes avec le statut `delivered`.

Ce choix permet d'analyser les ventes réellement finalisées et d'éviter d'inclure les commandes annulées, indisponibles ou encore en cours de traitement.

```python
sales_delivered = sales_clean[
    sales_clean["order_status"] == "delivered"
].copy()

sales_delivered.shape
```

### Vérification des KPIs de base

Avant d'exporter les données propres, nous calculons quelques indicateurs simples afin de vérifier que la table `sales_delivered` est cohérente.

Ces premiers KPIs permettent de contrôler rapidement le volume des ventes, le chiffre d'affaires et le panier moyen.

```python
total_revenue = sales_delivered["revenue"].sum()
total_orders = sales_delivered["order_id"].nunique()
total_customers = sales_delivered["customer_unique_id"].nunique()
average_order_value = total_revenue / total_orders
total_items_sold = len(sales_delivered)

basic_kpis = pd.DataFrame({
    "KPI": [
        "Total revenue",
        "Total orders",
        "Total customers",
        "Average order value",
        "Total items sold"
    ],
    "Value": [
        round(total_revenue, 2),
        total_orders,
        total_customers,
        round(average_order_value, 2),
        total_items_sold
    ]
})

basic_kpis
```

### Export des données propres

Après nettoyage et enrichissement, nous exportons les tables propres au format CSV.

Ces fichiers seront utilisés dans les prochaines étapes du projet :

- requêtes SQL ;
- dashboard Power BI ;
- analyses complémentaires ;
- publication GitHub.

```python
CLEAN_PATH = "../data/clean/"

sales_clean.to_csv(
    os.path.join(CLEAN_PATH, "sales_clean.csv"),
    index=False
)

sales_delivered.to_csv(
    os.path.join(CLEAN_PATH, "sales_delivered.csv"),
    index=False
)

basic_kpis.to_csv(
    os.path.join(CLEAN_PATH, "basic_kpis.csv"),
    index=False
)

print("Exports terminés avec succès.")
```
